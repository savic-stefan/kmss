```python
import numpy as np

# Создаем матрицу коэффициентов
A = np.array([[2, 1, 4],
              [3, 2, 1],
              [1, 3, 3]])

# Создаем вектор правой части уравнений
b = np.array([16, 10, 16])

# Применяем метод Гаусса
x = np.linalg.solve(A, b)

# Выводим решение
print("Решение: ")
print("x1 =", x[0])
print("x2 =", x[1])
print("x3 =", x[2])
```

Задание:
Запрограммировать модель Хищник-жертва и провести численные
эксперименты. Построить график полученных результатов.

```python
import numpy as np
import matplotlib.pyplot as plt

# Коэффициенты модели
alpha = 0.1
beta = 0.02
gamma = 0.3
delta = 0.01

# Начальные значения популяций
x0 = 100  # Жертвы
y0 = 15   # Хищники

# Временные параметры
dt = 0.001  # Шаг времени
t_max = 100  # Максимальное время моделирования

# Массивы для хранения результатов
t_values = np.arange(0, t_max, dt)
x_values = np.zeros_like(t_values)
y_values = np.zeros_like(t_values)

# Инициализация начальных значений
x_values[0] = x0
y_values[0] = y0

# Численное решение системы уравнений Эйлера
for i in range(1, len(t_values)):
    dx_dt = alpha * x_values[i-1] - beta * x_values[i-1] * y_values[i-1]
    dy_dt = delta * x_values[i-1] * y_values[i-1] - gamma * y_values[i-1]
    x_values[i] = x_values[i-1] + dt * dx_dt
    y_values[i] = y_values[i-1] + dt * dy_dt

# Построение графика
plt.plot(t_values, x_values, label='Жертвы')
plt.plot(t_values, y_values, label='Хищники')
plt.xlabel('Время')
plt.ylabel('Популяция')
plt.legend()
plt.show()
```
